### only muc - berlin 

```python
# 1. find all ride_ids where the ride touches a München station
muc_ride_ids = df.loc[df["station_name"].str.contains("München", na=False), "train_line_ride_id"]

# 2. find all ride_ids where the ride touches a Berlin station
berlin_ride_ids = df.loc[df["station_name"].str.contains("Berlin", na=False), "train_line_ride_id"]

# 3. keep only ICE rides that touch both
target_ids = set(muc_ride_ids) & set(berlin_ride_ids)

muc_berlin_stops = df[
    (df["train_line_ride_id"].isin(target_ids))
].copy()

print(muc_berlin_stops["train_line_ride_id"].nunique(), "unique ride IDs")
```

```python
muc_berlin_stops["date"] = muc_berlin_stops["time"].dt.date

connections_per_day = muc_berlin_stops.groupby("date")["train_line_ride_id"].nunique().reset_index(name="n_connections")
print(connections_per_day)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 5))
plt.plot(connections_per_day["date"], connections_per_day["n_connections"], marker="o", markersize=2)
plt.xlabel("Date")
plt.ylabel("Number of connections")
plt.title("MUC–Berlin ICE connections per day")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

daily_delay = muc_berlin_stops.groupby("date")["delay_in_min"].sum().reset_index(name="total_delay_min")

daily_delay.plot(x="date", y="total_delay_min", figsize=(14, 4), title="Accumulated delay per day (MUC–Berlin)")
plt.ylabel("Total delay (min)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# extract temperature as a dataframe
t2m_df = t2m_dtl["tg"].to_dataframe().reset_index()
t2m_df["date"] = t2m_df["time"].dt.date

# daily delay (as before)
daily_delay = muc_berlin_stops.groupby("date")["delay_in_min"].sum().reset_index(name="total_delay_min")
daily_delay["date"] = pd.to_datetime(daily_delay["date"])
t2m_df["date"] = pd.to_datetime(t2m_df["date"])

# restrict temperature to the same period as the delay data
t2m_df = t2m_df[
    (t2m_df["date"] >= daily_delay["date"].min()) &
    (t2m_df["date"] <= daily_delay["date"].max())
]
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

ax1.plot(daily_delay["date"], daily_delay["total_delay_min"])
ax1.set_ylabel("Total delay (min)")
ax1.set_title("Accumulated delay per day (MUC–Berlin)")

ax2.plot(t2m_df["date"], t2m_df["tg"], color="tab:orange")
ax2.set_ylabel("Mean temp (°C)")
ax2.set_title("Mean daily temperature over Germany (E-OBS)")
ax2.set_xlabel("Date")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

merged = pd.merge(t2m_df[["date", "tg"]], daily_delay, on="date", how="inner")

plt.figure(figsize=(8, 6))
plt.scatter(merged["total_delay_min"], merged["tg"], alpha=0.5, s=15)
plt.xlabel("Total delay (min)")
plt.ylabel("Mean daily temperature (°C)")
plt.title("Delay vs. temperature (MUC–Berlin)")
plt.tight_layout()
plt.show()
```
